# Phase 1 — Business Understanding (CRISP-DM)

This notebook defines the business problem, objectives, stakeholders, success criteria, and ethical boundaries for the Drug Review text mining project. No model training is performed here.

## 1. Application Domain

Patients share free-text experiences about prescription and over-the-counter drugs on platforms such as Druglib.com. These reviews describe **perceived benefits** and **side effects** in natural language, alongside structured labels such as effectiveness and side-effect severity.

Healthcare and pharmacovigilance teams need scalable ways to turn this unstructured text into actionable signals: which reviews suggest high/low effectiveness, and which side-effect themes appear most often across the corpus.

**Dataset:** [UCI Drug Review Dataset (Druglib.com)](https://archive.ics.uci.edu/dataset/461/drug+review+dataset+druglib+com) — patient reviews with fields including `benefitsReview`, `sideEffectsReview`, `effectiveness`, `sideEffects`, `condition`, `urlDrugName`, and `rating`.

## 2. Business Problem Statement

**Problem:** Manually reading thousands of drug reviews to assess effectiveness sentiment and side-effect themes does not scale, is inconsistent across reviewers, and delays insight for product safety and patient-support teams.

**How text mining helps:**
1. **Supervised classification (Model 1):** Predict a drug’s labeled **effectiveness** category from the patient’s `benefitsReview` text (TF-IDF + linear classifiers), comparing a fine-grained **5-class** scheme with a coarser **3-class** scheme.
2. **Unsupervised topic modeling (Model 2):** Discover latent **side-effect themes** from `sideEffectsReview` text using LDA, then manually label topics from top keywords.

Together, these models support triage: flag reviews that sound ineffective or severe, and group side-effect language into interpretable themes for dashboards and human review.

## 3. Project Objectives & Expected Outcomes

| Objective | Expected outcome |
|-----------|------------------|
| Predict effectiveness from benefits text | Trained LR / Linear SVM models with metrics vs a majority-class baseline |

| Stakeholder | Interest |
|-------------|----------|

| **Clinicians / pharmacists** | Fast scan of benefit language and recurring adverse-effect themes (decision support only) |

## 5. Why This Problem Matters

- Effectiveness and side effects are the two axes patients care about most when choosing or continuing a therapy.
- Structured ratings alone miss nuance (“helped my pain but ruined my sleep”). Free text carries that nuance.
- Volume of online reviews grows faster than human review capacity; automated first-pass analysis reduces backlog.
- Topic themes make side-effect language searchable and reportable without requiring every review to use the same vocabulary.

## 6. Success Criteria

**Model 1 (classification)**
- Beat a majority-class baseline on held-out test data (accuracy and especially **macro-F1**, given imbalance).
- Produce interpretable evidence (e.g., top TF-IDF terms / LR coefficients per class).
- Quantify the trade-off between **5-class** granularity and **3-class** reliability.

**Model 2 (topics)**
- Extract a coherent set of 5–8 side-effect topics.
- Each topic receives a manual label grounded in top keywords.
- Sample reviews assigned to topics should be qualitatively plausible.

**Process**
- Transparent preprocessing, fixed random seeds where applicable, and artifacts saved under `data/processed/` and `visuals/`.

## 7. Label Schemes for Model 1

**5-class (original):** Highly Effective · Considerably Effective · Moderately Effective · Marginally Effective · Ineffective

**3-class (binned for comparison):**

| 3-class | Source 5-class labels |
|---------|------------------------|
| **High** | Highly Effective, Considerably Effective |
| **Moderate** | Moderately Effective |
| **Low** | Marginally Effective, Ineffective |

Rationale: the original distribution is skewed toward Highly/Considerably Effective; binning reduces adjacent-class confusion while preserving a clinically meaningful High / Moderate / Low distinction.

## 8. Ethical Considerations

- **Not medical advice:** Models support research and triage; they must not recommend or discontinue medication.
- **Observational bias:** Reviews are self-selected; dissatisfied or highly motivated patients may over-represent extremes.
- **Class imbalance:** Rare classes (e.g., Ineffective) need balanced class weights and macro-averaged metrics so majority classes do not dominate evaluation.
- **Missing demographics:** Age, sex, dose, and comorbidities are largely absent — predictions can encode unobserved confounders.
- **Privacy & license:** Dataset is for non-commercial research use (UCI terms); do not redistribute raw reviews in public forks without compliance.
- **Human-in-the-loop:** Severe side-effect themes should route to human reviewers, not automated clinical action.
